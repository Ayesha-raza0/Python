```python
#comparision operator
N1 = 100
N2 = 150
print(N1 == N2) #checks if values are same or not
print(N1 > N2) #greater or not
print(N1 < N2) #less or not
print(N1 >= N2) #greater or equal
print(N1 <= N2) #less or equal
print(N1 != N2) #notequal to
```

#LIST METHOD


```python
veg = ["potato","tomato","pease","cucumber"]
veg.append("alu palak")
print(veg)
veg.insert(3,"gobi")
print(veg)
veg.sort(reverse=False)
print("sorted list",veg)
veg.sort(reverse=True)
print(veg)
print("pease count",veg.count("pease"))
veg_copy = veg.copy()
print(veg.copy)
veg_copy.clear()
print("cleared list:", veg_copy)
print(veg.index("pease"))
```
