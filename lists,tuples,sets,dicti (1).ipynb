#SETS

```python
veg = {"potato","tomato","pease","cucumber","potato"}
print(veg)
#sets method
set1 = {1,2,3,4,5}
set2 = {5,6,7,8,9,10}
nums = [20,30,40]
set1.add(15)
print(set1)
set2.update(nums)
print(set2)
```

```python
print(set1.intersection(set2)) #same element
print(set1.union(set2)) # 1 tarteeb main likh dega donon set
print(set1.difference(set2))#wo element jo set 2 main nhi h
print(set2.difference(set1))
print(set1.isdisjoint(veg)) #kuch bhi common nhi tou true ajayga
print(set2.symmetric_difference(set1)) #ascending order main likh dega
set2.clear()
print("cleared",set2)
set3 = {1,2,3}
set4 = {1,2,3,4,5,6,7,8,9}
set5 = {1,2,3,4,5}
print(set4.issuperset(set3)) #set4 is bigger set tou true ayga
print(set3.issubset(set4)) #set3 is a subset of set4 tou true ayga
print(set3.issubset(set5)) 
veg.discard("cucumber")
print("Discarded",veg)


```
