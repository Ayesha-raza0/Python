```python
#comparision operator
N1 = 100
N2 = 150
print(N1 == N2) #checks if values are same or not
print(N1 > N2) #greater or not
print(N1 < N2) #less or not
print(N1 >= N2) #greater or equal
print(N1 <= N2) #less or equal
print(N1 != N2) #notequal to
```
