Question 2 (5 Marks) – Coding
∙ Write a program to manage student marks using lists and
dictionaries.

```python
#i) Input details of 5 students (name and marks).
name1 = input("Enter name of student 1  : ")
marks1 = float(input("Enter your Marks :"))
 
name2 = input("Enter name of student 2  : ")
marks2 = float(input("Enter your Marks :"))
 
name3 = input("Enter name of student 3  ")
marks3 = float(input("Enter your Marks :"))
 
name4 = input("Enter name of student 4  ")
marks4 = float(input("Enter your Marks :"))
 
name5 = input("Enter name of student 5  ")
marks5 = float(input("Enter your Marks :"))
 #ii) Store them in a dictionary.
students = {
    name1: marks1,
    name2: marks2,
    name3: marks3,
    name4: marks4,
    name5: marks5
}
print(students)
#iii)print the list using for loop
for name, marks in students.items():
    print(name, marks)
#iv) Display highest and lowest marks. (use max and min methods)
highest = max(students.values())
lowest = min(students.values())

print("Highest Marks:", highest)
print("Lowest Marks:", lowest)
```

```python
name = "Hello iam ayesha raza"

# 1. capitalize
print(name.capitalize())

# 2. lower()
print(name.lower())

# 3. replace()
print(name.replace("raza", "khan"))

# 4. split()
print(name.split())

# 5. len()
print(len(name))
```

Question 4 (5 Marks) – Coding
∙ A class has students enrolled in different clubs.
∙ i) Create a nested dictionary where each club stores a list of student
names

- ii) Create 2 new sets having club members
∙ iii) Find students common to two clubs using sets made above in last part.
∙ iv) Ask the user for a club name and display its members using conditional
statements.

```python
# i) Create a nested dictionary
clubs = {
    "Sports": ["Ali", "Sara", "Ahmed"],
    "Music": ["Sara", "Ayesha", "Usman"],
    "Drama": ["Ahmed", "Ayesha", "Hina"]
}

print("Clubs:", clubs)

# ii) Create 2 sets having club members
sports = set(clubs["Sports"])
music = set(clubs["Music"])

print("Sports Set:", sports)
print("Music Set:", music)

# iii) Find students common to two clubs
common = sports.intersection(music)

print("Common Students:", common)

# iv) Ask user for a club name and display its members
club_name = input("Enter club name: ")

if club_name == "Sports":
    print("Sports Members:", clubs["Sports"])

elif club_name == "Music":
    print("Music Members:", clubs["Music"])

elif club_name == "Drama":
    print("Drama Members:", clubs["Drama"])

else:
    print("Club not found")

```
