```python
# calculator
number1 = float(input("Enter any number"))
operator = input("Enter any operator:")
number2 = float(input("Enter any number"))
if operator == "+":
    print(number1 + number2)
elif operator == "-":
    print(number1 - number2)
elif operator == "/":
    if number2 == 0:
        print("Number cannot be divided by 0")
    else:
        print(number1 / number2)
elif operator == "x":
    print(number1 * number2)
elif operator == "^":
    print(number1 ** number2)
elif operator == "%":
    print(number1 % number2)
elif operator == "//":
    if number2 == 0:
        print("Number cannot be divided by 0")
    else:
        print(number1 // number2)
else:
    print("Invalid operator")
```

```python
# nested if-else 
# temp < 20 -> cold, 20 -30 -> pleasant, 31-40 Hot, 40> very hot
is_raining = True 
temperature = int(input("enter any number: ")) 
if is_raining == True:
    print("It is raining outside")
    if temperature < 20:#nested start 
        print("Cold")
    elif temperature >=20 and temperature <30: 
        print("Pleasant") 
    elif temperature >30 and temperature < 40: 
        print("hot") 
    else: print("very hot") #nested end else: print("It is not raining outside")
```
